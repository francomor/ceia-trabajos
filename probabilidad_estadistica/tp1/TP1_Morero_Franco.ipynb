1. (2 puntos) De 10 monedas hay 5 monedas falsas, que tienen probabilidad $0{,}3$ de mostrar cara al ser lanzadas. Si se toma una moneda al azar, se lanza 16 veces, y en todas ellas se obtiene cara, ¿qué es más probable, que la moneda elegida sea justa o que esté cargada? Dar las probabilidades tanto de que la moneda elegida sea falsa, como de que sea justa.

---

Tenemos $10$ monedas, de ellas $5$ falsas.

Definimos:

$A$: moneda verdadera $\rightarrow$ Probabilidad de sacar cara: $0.5$, porque no esta cargada.

$B$: moneda falsa $\rightarrow$ Probabilidad de sacar cara: $0.3$.

$C$: obtener $16$ caras seguidas.

Entonces lo que hacemos es de la "bolsa de monedas", tomar una al azar, y con eso la tiramos $16$ veces y en todas nos salio cara.

Podemos ver entonces:

$$
P(A) = \frac{5}{10} = 0.5 \qquad P(B) = \frac{5}{10} = 0.5
$$

Como los 16 tiros son independientes, para calcular la probabilidad de que saquemos cara 16 veces, dado que tomamos un tipo de moneda, lo que vamos a hacer es multiplicar la probabilidad de sacar cara para el tipo de moneda 16 veces. Vemos que:


$P(C|A) = 0.5^{16}$

$P(C|B) = 0.3^{16}$

Luego por la ley de probabilidad total:

$$
P(C) = P(C|A)\,P(A) + P(C|B)\,P(B) = 0.5^{16} \cdot 0.5 + 0.3^{16} \cdot 0.5
$$

Usando el teorema de Bayes:

$$
P(A|C) = \frac{P(C|A)\,P(A)}{P(C)}
$$

$$
P(B|C) = \frac{P(C|B)\,P(B)}{P(C)}
$$

```python
p_a_dado_c = 0.5**16 / (0.5**16 + 0.3**16)
p_b_dado_c = 0.3**16 / (0.5**16 + 0.3**16)
print(f"P(A|C) = {p_a_dado_c*100:.2f}%")
print(f"P(B|C) = {p_b_dado_c*100:.2f}%")
```

Podemos ver entonces que es muchisimo mas probable que la moneda elegida sea verdadera ($A$), dado que:

Probabilidad de moneda verdadera: $P(A|C) \approx 99.97\%$

Probabilidad de moneda falsa: $P(B|C) \approx 0.03\%$

2. (2 puntos) Sean $X$ e $Y$ dos v.a. continuas con densidad conjunta:

$$
f_{X,Y} =
\begin{cases}
Ky & 25x^2 \le y \le 7x \\
0 & \text{en otro caso}
\end{cases}
$$

Encontrar:

a) Determine el valor de $K$.

b) Encuentre la densidad marginal $f_X(x)$ de $X$.

c) Encuentre la densidad marginal $f_Y(y)$ de $Y$.

d) Encuentre la densidad condicional $f_{X|Y}(x|y)$ de $X$ dado $Y$.

e) Encuentre la densidad condicional $f_{Y|X}(y|x)$ de $Y$ dado $X$.

---

a) Determinar el valor de $K$


La densidad conjunta es:

$$
f_{X,Y}(x,y)=Ky
$$

para la región:

$$
25x^2 \le y \le 7x
$$

Primero buscamos los puntos de corte:

$$
25x^2 = 7x
$$

$$
25x^2 - 7x = 0
$$

$$
x(25x-7)=0
$$

Por lo tanto:

$$
x=0
$$

o bien:

$$
x=\frac{7}{25}
$$

Entonces, para que sea una densidad:

$$
\int_0^{7/25}\int_{25x^2}^{7x} Ky\,dy\,dx = 1
$$

Integramos:

$$
K\int_0^{7/25}
\left[
\frac{y^2}{2}
\right]_{25x^2}^{7x}
dx = 1
$$

$$
K\int_0^{7/25}
\left(
\frac{(7x)^2}{2}
-
\frac{(25x^2)^2}{2}
\right)
dx = 1
$$

$$
\frac{K}{2}\int_0^{7/25}
\left(
49x^2 - 625x^4
\right)
dx = 1
$$

$$
\frac{K}{2}
\left[
\frac{49x^3}{3}
-
\frac{625x^5}{5}
\right]_0^{7/25}
= 1
$$

$$
\frac{K}{2}
\left[
\frac{49x^3}{3}
-
125x^5
\right]_0^{7/25}
= 1
$$

$$
\frac{K}{2}
\left(
\frac{49(7/25)^3}{3}
-
125(7/25)^5
\right)
= 1
$$

```python
expr = (49*(7/25)**3)/3 - 125*(7/25)**5
K = 2 / expr
print(f"K = {K:.4f}")
```

Por aproximación, tomamos $K=14$ para simplificar cuentas:

$$
K \approx 14
$$

c) Densidad marginal de $Y$


Tenemos que buscar los limites de integracion en y:

$$
y = 7x
$$

$$
y = 25x^2
$$

Despejamos $x$:

Para $y = 7x$:

$$
x = \frac{y}{7}
$$

Para $y = 25x^2$:

$$
x = \sqrt{\frac{y}{25}}
$$

$$
x = \frac{\sqrt{y}}{5}
$$

Entonces:

$$
f_Y(y)=\int_{y/7}^{\sqrt{y}/5} 14y \, dx
$$

$$
= 14y \cdot x \Bigg|_{y/7}^{\sqrt{y}/5}
$$

$$
= 14y \cdot \frac{\sqrt{y}}{5} - 14y \cdot \frac{y}{7}
$$

$$
= \frac{14}{5} y^{3/2} - 2y^2
$$

Definimos dónde se mueve $Y$.

Sabemos que arranca en $x=0$:

$$
y_{\min}= 0
$$

y si tomamos $x = {\frac{7}{25}}$

$$
y_{\max}=7x = 7\left(\frac{7}{25}\right)=\frac{49}{25}
$$

Entonces:

$$
f_Y(y)=
\begin{cases}
\frac{14}{5}y^{3/2}-2y^2,
& 0 \le y \le \frac{49}{25} \\[10pt]
0,
& \text{en otro caso}
\end{cases}
$$

3. (2 puntos) Sea:

$$
Y = R\cos(2\pi\,\Phi),
$$

donde $R$ y $\Phi$ son variables aleatorias independientes. La fase $\Phi$ se distribuye uniforme sobre el intervalo $(0,1)$, mientras que la amplitud $R$ tiene una distribución de Rayleigh con densidad $f(r) = r\,e^{-r^2/2}$ para $r > 0$. Muestre que $Y$ tiene una distribución normal.

Pista: Defina $Y = R\,\operatorname{sen}(2\pi\,\Phi)$ y $X = R\cos(2\pi\,\Phi)$ y utilice la relación $X^2 + Y^2 = R^2$ para calcular la densidad conjunta de $(X,Y)$ a partir de la densidad conjunta de $(R,\Phi)$.

---

Sea

$$
Y = R \cos(2\pi \Phi)
$$

con $R$ y $\Phi$ v.a. independientes.

La variable $\Phi$ tiene distribución uniforme en $(0,1)$:

$$
f_{\Phi}(\phi)=\frac{1}{1-0}=1, \qquad 0<\phi<1
$$

y $R$ tiene densidad

$$
f_R(r)=r\,e^{-r^2/2}, \qquad r>0
$$

Como $R$ y $\Phi$ son independientes:

$$
f_{R,\Phi}(r,\phi)=f_R(r)\,f_{\Phi}(\phi)
$$

por lo tanto,

$$
f_{R,\Phi}(r,\phi)=r\,e^{-r^2/2},
\qquad r>0,\;0<\phi<1
$$


Sabemos de coordenadas polares que

$$
x=r\sin(\theta)
$$

$$
y=r\cos(\theta)
$$

y además

$$
r^2=x^2+y^2
$$

$$
dx\,dy=r\,dr\,d\theta
$$

Ahora, en nuestro caso:

$$
X=R\sin(2\pi\Phi)
$$

$$
Y=R\cos(2\pi\Phi)
$$

Vemos que nuestro angulo segun definición es

$$
\theta=2\pi\Phi
$$

entonces

$$
d\theta=2\pi\,d\Phi
$$

y por lo tanto

$$
dx\,dy = r\,dr\,d\theta = r\,dr\,(2\pi\,d\Phi)
$$

es decir,

$$
dx\,dy = 2\pi\,r\,dr\,d\Phi
$$

Como la probabilidad tiene que ser igual sin importar las coordenadas:

$$
f_{X,Y}(x,y)\,dx\,dy = f_{R,\Phi}(r,\phi)\,dR\,d\phi
$$

Entonces:

$$
f_{X,Y}(x,y)\,(2\pi\,r\,dr\,d\phi) = f_{R,\Phi}(r,\phi)\,dr\,d\phi
$$

Despejando:

$$
f_{X,Y}(x,y) = \frac{1}{2\pi r}\,f_{R,\Phi}(r,\phi)
$$

Reemplazando $f_{R,\Phi}$:

$$
f_{X,Y}(x,y) = \frac{1}{2\pi r}\left(r\,e^{-r^2/2}\right)
$$

$$
f_{X,Y}(x,y) = \frac{1}{2\pi}\,e^{-r^2/2}
$$

Como $r^2=x^2+y^2$ queda:

$$
f_{X,Y}(x,y) = \frac{1}{2\pi}\,e^{-(x^2+y^2)/2}
$$


Ahora como queremos ver que $Y$ tiene una distribución normal:

$$
f_Y(y) = \int_{-\infty}^{\infty} f_{X,Y}(x,y)\,dx
$$

$$
f_Y(y) = \int_{-\infty}^{\infty} \frac{1}{2\pi}\,e^{-(x^2+y^2)/2}\,dx
$$

Sacamos afuera de lo que no depende de $X$:

$$
f_Y(y) = \frac{1}{2\pi}\,e^{-y^2/2}\int_{-\infty}^{\infty} e^{-x^2/2}\,dx
$$

Sabemos que la integral gaussiana vale:

$$
\int_{-\infty}^{\infty} e^{-x^2/2}\,dx = \sqrt{2\pi}
$$

Entonces:

$$
f_Y(y) = \frac{1}{2\pi}\,e^{-y^2/2}\,\sqrt{2\pi}
$$

Como $2\pi = \sqrt{2\pi}\,\sqrt{2\pi}$ obtenemos:

$$
f_Y(y) = \frac{1}{\sqrt{2\pi}}\,e^{-y^2/2}
$$


Por definición, una normal $N(\mu,\sigma)$ tiene densidad

$$
f(y) = \frac{1}{\sigma\sqrt{2\pi}}\,e^{-(y-\mu)^2/(2\sigma^2)}
$$

Tomando $\mu=0$ y $\sigma=1$:

$$
f_Y(y) = \frac{1}{\sqrt{2\pi}}\,e^{-y^2/2}
$$

que coincide exactamente con la densidad obtenida.

Por lo tanto:

$$
Y \sim N(0,1)
$$

a) Simulación de clientes diarios usando Poisson


Para cada fecha $t$ entre $2023$-$01$-$01$ y $2025$-$12$-$31$, el parámetro Poisson es:

$$
\lambda_t = e_{\text{anual}} + e_{\text{mensual}} + e_{\text{diario}} + e_{\text{local}}
$$

y el número de clientes se simula como $\text{Poisson}(\lambda_t)$ de manera independiente para Don Francisco y Don Miguel.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
efecto_anual = {
    2023: 10, 
    2024: 15, 
    2025: 20
}

efecto_mensual = {
    1: 10, 
    2: 15, 
    3: 20, 
    4: 20, 
    5: 25, 
    6: 25,
    7: 30, 
    8: 25, 
    9: 25, 
    10: 20, 
    11: 15, 
    12: 10,
}

# pandas dayofweek: Lunes=0, ..., Domingo=6
efecto_diario = {
    0: 20,  # Lunes
    1: 30,  # Martes
    2: 35,  # Miércoles
    3: 30,  # Jueves
    4: 20,  # Viernes
    5: 10,  # Sábado
    6: 10,  # Domingo
}

efecto_local = {
    "Francisco": 20, 
    "Miguel": 10
}

fechas = pd.date_range("2023-01-01", "2025-12-31", freq="D")

df = pd.DataFrame({"fecha": fechas})
df["anio"] = df["fecha"].dt.year
df["mes"] = df["fecha"].dt.month
df["dia_semana"] = df["fecha"].dt.dayofweek

df["e_anual"] = df["anio"].map(efecto_anual)
df["e_mensual"] = df["mes"].map(efecto_mensual)
df["e_diario"] = df["dia_semana"].map(efecto_diario)

df["lambda_francisco"] = df["e_anual"] + df["e_mensual"] + df["e_diario"] + efecto_local["Francisco"]
df["lambda_miguel"]    = df["e_anual"] + df["e_mensual"] + df["e_diario"] + efecto_local["Miguel"]

rng = np.random.default_rng(42)
df["clientes_francisco"] = rng.poisson(df["lambda_francisco"])
df["clientes_miguel"]    = rng.poisson(df["lambda_miguel"])

df.head(1000)
```

---

b) Distribución empírica y densidad aproximada para Don Francisco

```python
anios = [2023, 2024, 2025]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for i, anio in enumerate(anios):
    data = df.loc[df["anio"] == anio, "clientes_francisco"]

    # ECDF (Distribución Acumulada Empírica)
    sns.ecdfplot(data, ax=axes[0, i])
    axes[0, i].set_title(f"Distribución acumulada empírica Don Francisco — {anio}")
    axes[0, i].set_xlabel("Clientes diarios")
    axes[0, i].set_ylabel(r"$\hat{F}_n(x)$")

    # Histograma + KDE (Estimación de Densidad)
    sns.histplot(data, bins=30, stat="density", alpha=0.4, ax=axes[1, i], label="Histograma")
    sns.kdeplot(data, ax=axes[1, i], color="red", lw=2, label="KDE")
    axes[1, i].set_title(f"Densidad aproximada Don Francisco — {anio}")
    axes[1, i].set_xlabel("Clientes diarios")
    axes[1, i].set_ylabel("Densidad")
    axes[1, i].legend()

plt.tight_layout()
plt.show()
```

---

c) Distribución empírica y densidad aproximada para Don Miguel

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for i, anio in enumerate(anios):
    data = df.loc[df["anio"] == anio, "clientes_miguel"]

    # ECDF (Distribución Acumulada Empírica)
    sns.ecdfplot(data, ax=axes[0, i])
    axes[0, i].set_title(f"Distribución acumulada empírica Don Miguel — {anio}")
    axes[0, i].set_xlabel("Clientes diarios")
    axes[0, i].set_ylabel(r"$\hat{F}_n(x)$")

    # Histograma + KDE (Estimación de Densidad)
    sns.histplot(data, bins=30, stat="density", alpha=0.4, ax=axes[1, i], label="Histograma")
    sns.kdeplot(data, ax=axes[1, i], color="red", lw=2, label="KDE")
    axes[1, i].set_title(f"Densidad aproximada Don Miguel — {anio}")
    axes[1, i].set_xlabel("Clientes diarios")
    axes[1, i].set_ylabel("Densidad")
    axes[1, i].legend()

plt.tight_layout()
plt.show()
```

d) Densidad empírica conjunta para ambos almacenes

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for i, anio in enumerate(anios):
    sub = df.loc[df["anio"] == anio]

    # KDE 2D
    sns.kdeplot(
        data=sub, x="clientes_francisco", y="clientes_miguel",
        fill=True, cmap="Blues", levels=20, thresh=0.02,
        ax=axes[i],
    )
    axes[i].set_title(f"KDE conjunta — {anio}")
    axes[i].set_xlabel("Clientes Don Francisco")
    axes[i].set_ylabel("Clientes Don Miguel")

plt.tight_layout()
plt.show()
```
